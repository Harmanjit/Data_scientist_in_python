```python
import sqlite3 as sql
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Path to .db file from Google Takeout
db_path = "raw_data/health_connect_export.db"  

# Connect to the database
conn = sql.connect(db_path)
```

```python
# Get a list of all tables in the database
query = "SELECT name FROM sqlite_master WHERE type='table';"
tables = pd.read_sql(query, conn)
# pd.set_option('display.max_rows', None)  # Show all rows
# pd.set_option('display.max_columns', None)  # Show all columns
print("Available tables in the database:")
print(tables)
```

```python
# If there's a table related to glucose monitoring, let's say it's named "glucose_data"
# Replace 'glucose_data' with the actual table name after inspecting the available tables
df = pd.read_sql("SELECT * FROM blood_glucose_record_table", conn)

# Close the connection
# conn.close()

# Display the first few rows of the data
print(df.head())
```

How to convert this into mg/dL, which is more interpretable. 
I know my average glucose levels are 104mg/mL 
SO average of this will be asigned 104mg/mL

```python
df.level.mean()
# df.level.median()
```

```python
df.level.min()
```

```python
# Known original statistics
mu_orig = df['level'].mean()  # Original mean
L_orig = df['level'].min()    # Original lower bound

# Desired statistics
mu_desired = 104  # New mean
L_desired = 70  # New lower bound
```

```python
# df['level_scaled'] = ((df['level'] - mu_orig) * ((V_desired - L_desired) / (mu_orig - L_orig))) + L_desired
df['level_scaled'] = L_desired + ((df['level'] - L_orig) / (mu_orig - L_orig)) * (mu_desired - L_desired)
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(x=df['local_date_time'], y=df['level_scaled'], marker='o', linestyle='-')
plt.xlabel("Date")
plt.ylabel("level")
plt.title("Glucose levels")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
